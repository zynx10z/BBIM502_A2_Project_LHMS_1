```python
def add_room(self, room_no, room_type):
    if room_no in self.rooms:
        print("Room already exists.")
    else:
        self.rooms[room_no] = room_type
        print(f"Room {room_no} added successfully.")

def display_rooms(self):
    if not self.rooms:
        print("No rooms available.")
    else:
        for room_no, room_type in self.rooms.items():
            print(f"Room Number: {room_no}")
            print(f"Room Type: {room_type}")
```

```python
def menu():
    hotel = HotelManagement()

    while True:
        print("\n1. Add Room")
        print("2. Display Rooms")
        print("0. Exit")

        choice = input("Enter Choice: ")

        if choice == "1":
            room_no = input("Room Number: ")
            room_type = input("Room Type: ")
            hotel.add_room(room_no, room_type)

        elif choice == "2":
            hotel.display_rooms()

        elif choice == "0":
            break

menu()
```
