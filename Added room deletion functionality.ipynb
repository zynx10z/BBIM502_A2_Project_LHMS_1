```python
class HotelManagement:
    def __init__(self):
        self.rooms = {}

    # ADD ROOM (Day 1)
    def add_room(self, room_no, room_type):
        if room_no in self.rooms:
            print("Room already exists.")
        else:
            self.rooms[room_no] = room_type
            print(f"Room {room_no} added successfully.")

    # DISPLAY ROOMS (Day 1)
    def display_rooms(self):
        if not self.rooms:
            print("No rooms available.")
        else:
            for room_no, room_type in self.rooms.items():
                print(f"Room Number: {room_no}")
                print(f"Room Type: {room_type}")

    # DELETE ROOM (DAY 2 ADDITION)
    def delete_room(self, room_no):
        if room_no in self.rooms:
            self.rooms.pop(room_no)
            print(f"Room {room_no} deleted successfully.")
        else:
            print("Room not found.")
```

```python
def menu():
    hotel = HotelManagement()

    while True:
        print("\n==============================")
        print("LANGHAM HOTEL MANAGEMENT SYSTEM")
        print("==============================")
        print("1. Add Room")
        print("2. Delete Room")
        print("3. Display Rooms")
        print("0. Exit")
        print("==============================")

        choice = input("Enter Choice: ")

        if choice == "1":
            room_no = input("Room Number: ")
            room_type = input("Room Type: ")
            hotel.add_room(room_no, room_type)

        elif choice == "2":
            room_no = input("Enter Room Number to Delete: ")
            hotel.delete_room(room_no)

        elif choice == "3":
            hotel.display_rooms()

        elif choice == "0":
            print("Program Closed.")
            break

        else:
            print("Invalid choice. Try again.")

menu()
```
